# Maintenance Analysis Project

## Data Loading and Initial Inspection

```python
import pandas as pd

# Load the dataset (assuming 'maintenance_data.csv' is available in the environment)
try:
    df = pd.read_csv('maintenance_data.csv')
    print("Dataset loaded successfully.")
except FileNotFoundError:
    print("Error: 'maintenance_data.csv' not found. Please ensure the file is uploaded or the path is correct.")
    df = pd.DataFrame() # Create an empty DataFrame to avoid errors later if file not found

# Display the first 5 rows of the DataFrame
if not df.empty:
    display(df.head())
```

## Data Cleaning and Validation

```python
# Display DataFrame information to check data types and non-null counts
print("DataFrame Info:")
if not df.empty:
    df.info()
else:
    print("DataFrame is empty. Please ensure 'maintenance_data.csv' was loaded successfully.")
```

```python
# Check for missing values
print("\nMissing Values:")
if not df.empty:
    display(df.isnull().sum().to_frame(name='Missing Count'))
else:
    print("DataFrame is empty. Cannot check for missing values.")
```

```python
# Display descriptive statistics for numerical columns
print("\nDescriptive Statistics:")
if not df.empty:
    display(df.describe())
else:
    print("DataFrame is empty. Cannot display descriptive statistics.")
```

```python
# Convert 'Maintenance_Date' to datetime objects
if not df.empty:
    try:
        df['Maintenance_Date'] = pd.to_datetime(df['Maintenance_Date'])
        print("\n'Maintenance_Date' column converted to datetime.")
        display(df.head())
    except KeyError:
        print("Error: 'Maintenance_Date' column not found. Please check column name.")
    except Exception as e:
        print(f"Error converting 'Maintenance_Date' to datetime: {e}")
else:
    print("DataFrame is empty. Cannot convert 'Maintenance_Date' column.")
```

```python
# Check for duplicate rows
if not df.empty:
    duplicates = df.duplicated().sum()
    print(f"\nNumber of duplicate rows: {duplicates}")
    if duplicates > 0:
        print("Dropping duplicate rows...")
        df.drop_duplicates(inplace=True)
        print("Duplicate rows dropped.")
    else:
        print("No duplicate rows found.")
else:
    print("DataFrame is empty. Cannot check for duplicate rows.")
```
